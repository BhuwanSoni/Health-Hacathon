```python
import pandas as pd
import numpy as np
from geopy.distance import geodesic
from pymongo import MongoClient
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.preprocessing import StandardScaler
```

```python
def blood_type_compatible(recipient_blood, donor_blood):
    compatibility = {
        "O-": ["O-"],
        "O+": ["O-", "O+"],
        "A-": ["O-", "A-"],
        "A+": ["O-", "O+", "A-", "A+"],
        "B-": ["O-", "B-"],
        "B+": ["O-", "O+", "B-", "B+"],
        "AB-": ["O-", "A-", "B-", "AB-"],
        "AB+": ["O-", "O+", "A-", "A+", "B-", "B+", "AB-", "AB+"]
    }
    return 1 if donor_blood in compatibility.get(recipient_blood, []) else 0
```

```python
def calculate_distance(lat1, lon1, lat2, lon2):
    return geodesic((lat1, lon1), (lat2, lon2)).kilometers
```

```python
client = MongoClient('mongodb://localhost:27017/')
db = client['Database']

# Load collections
donors_collection = db['Donor']
recipients_collection = db['Reciptent']
```

```python
donors = pd.DataFrame(list(donors_collection.find({}, {'_id': 0})))
recipients = pd.DataFrame(list(recipients_collection.find({}, {'_id': 0})))

# Debug: Print column names to verify fields
print("Donor Columns:", donors.columns)
print("Recipient Columns:", recipients.columns)
```

```python
if "Blood_Type" in donors.columns:
    donors.rename(columns={"Blood_Type": "blood_type"}, inplace=True)

if "Recipient_Blood_Type" in recipients.columns:
    recipients.rename(columns={"Recipient_Blood_Type": "blood_type"}, inplace=True)
```

```python
if "location_lat" not in donors.columns:
    donors["location_lat"] = np.random.uniform(-90, 90, len(donors))

if "location_long" not in donors.columns:
    donors["location_long"] = np.random.uniform(-180, 180, len(donors))

if "Location_Lat" in recipients.columns and "Location_Long" in recipients.columns:
    recipients.rename(columns={"Location_Lat": "location_lat", "Location_Long": "location_long"}, inplace=True)
```

```python
features = []
labels = []
```

```python
for _, recipient in recipients.iterrows():
    for _, donor in donors.iterrows():
        # Compatibility features
        blood_compatibility = blood_type_compatible(recipient["blood_type"], donor["blood_type"])
        distance = calculate_distance(
            recipient["location_lat"], recipient["location_long"],
            donor["location_lat"], donor["location_long"]
        )
        urgency_mapping = {"High": 2, "Moderate": 1, "Low": 0}
        urgency_level = urgency_mapping.get(recipient["Urgency_Level"], 0)
        
        # Match label: 1 for a valid match, 0 otherwise
        is_match = 1 if (blood_compatibility == 1 and distance <= 500) else 0

        # Append features and label
        features.append([blood_compatibility, distance, urgency_level])
        labels.append(is_match)
```
